#### Notice that the estimated coefficient decreases as we consider higher lags. However, OLS estimators in this case are not reliable due to the presence of *autocorrelation*, which violates the assumption of independent errors in OLS regression. The OLS coefficient is derived from $\frac{\text{Cov}(X,Y)}{\text{Var}(X)}$, which becomes biased when *autocorrelation* is present.

```python
lags = range(1, 59) 
OLS_coeff = []

for shift_period in lags:
    GDP_t = GDP[shift_period:]
    GDP_t_minus_shift = GDP[:-shift_period]

    GDP_t_minus_shift_intercept = sm.add_constant(GDP_t_minus_shift)

    model = sm.OLS(GDP_t, GDP_t_minus_shift_intercept)
    result = model.fit()
    OLS_coeff.append(result.params[1])

plt.figure(figsize=(10, 5))
plt.stem(lags, OLS_coeff, basefmt="gray", markerfmt="o", linefmt="purple")
plt.axhline(0, color='red', linestyle='--', label="Zero Reference Line")

plt.xlabel("Lag Periods ($k$)")
plt.ylabel("OLS Coefficients ($\\beta_k$)")
plt.title("OLS Estimated Coefficients for Lagged GDP Values")
plt.legend()
plt.show()
```

## 3rd Approach: Using Autocorrelation Function (ACF)

### $\gamma_k = \frac{1}{n} \sum_{t=1}^{n-k} (x_t - \bar{x}) (x_{t+k} - \bar{x})$
- $\hat \gamma_k$ is the autocovariance estimate (biased).
### $\rho_k = \frac{\gamma_k}{\sigma^2}$

### $\tilde \gamma_k = \frac{1}{n-k} \sum_{t=1}^{n-k} (x_t - \bar{x}) (x_{t+k} - \bar{x})$
- $\tilde \gamma_k$ is the autocovariance estimate (unbiased).
### $\rho_k = \frac{\tilde \gamma_k}{\sigma^2}$

### Let's first compute and visualize the autocovariance values. Covariance indicates the relationship between two variables, specifically how they move together.

```python
np.cov(GDP[1:], GDP[:-1])[0,1]
```

```python
np.cov(GDP[2:], GDP[:-2])[0,1]
```

```python
np.cov(GDP[3:], GDP[:-3])[0,1]
```

```python
def autocovariance(x, lags):
    """Compute autocovariance for each lag."""
    n = len(x)
    mean_x = np.mean(x)
    autocov_values = []

    for lag in range(1, lags + 1):
        cov = np.sum((x[:-lag] - mean_x) * (x[lag:] - mean_x)) / (n)
        autocov_values.append(cov)

    return np.array(autocov_values)

lags = 59
autocov_manual = autocovariance(GDP, lags)

#print("Autocovariance values:", autocov_manual)
```

```python
plt.stem(autocov_manual)  
plt.xlabel("Lag")
plt.ylabel("Autocovariance")
plt.title("Autocovariance Function")
plt.show()
```

#### The reason the autocovariance function from the statsmodels library returns to zero at higher lags, while our defined version does not, lies in the distinction between the biased and unbiased versions of the autocovariance function. The difference arises from the denominator used in normalization.

#### If we choose to divide by 𝑛 instead of 𝑛−𝑘, the result will be similar to the autocovariance function computed by statsmodels. The biased version, which divides by 𝑛, ensures stability and provides a more consistent estimate, preventing extreme fluctuations at higher lags. This makes it particularly useful in spectral analysis and certain time series applications where smoother estimates are preferred.

**Example**: Let's say we have 60 observations with a mean of zero. Consider the following autocovariance computations: $\gamma_{59} = \frac{(10-0)(5-0)}{60} = 0.83$ but if we divide by $n-k$ we will have $\gamma_{59} = \frac{(10-0)(5-0)}{60-59} = 50$

### Defining the ACF manually:

```python
def manual_acf(x, lags):
    """Compute ACF manually to match statsmodels.acf()"""
    n = len(x)
    mean_x = np.mean(x)
    var_x = np.var(x)  
    acf_values = []

    for lag in range(1, lags + 1):
        cov = np.sum((x[:-lag] - mean_x) * (x[lag:] - mean_x)) / (n - lag)
        acf_values.append(cov / var_x)

    return np.array(acf_values)

lags = 59
ACF_manual = manual_acf(GDP, lags)
```

```python
plt.stem(range(1, lags + 1), ACF_manual)
plt.axhline(0, color="red", linestyle="--")
plt.xlabel("Lag")
plt.ylabel("ACF")
plt.title("Manual ACF")
plt.show()
```

#### The plot is similar to the autocovariance plot, but the difference lies in the scale. The ACF divides the autocovariance by the variance, normalizing it so that it remains between -1 and 1.

# Partial Autocorrelation Function (PACF)

There are many methods for estimating the PACF coefficient, but I will derive it using Ordinary Least Squares (OLS). We will also explore the Yule-Walker method implemented by the statsmodels PACF function.

*The difference between ACF and PACF is that the ACF accounts for the indirect effects of lagged values, while the PACF captures only the direct effects. For instance, if we aim to understand the relationship between current inflation and inflation from one and two periods ago, the ACF considers the influence of inflation from two periods ago on inflation from one period ago, which then affects current inflation. In contrast, the PACF directly assesses the impact of inflation from two periods ago on current inflation, disregarding any intermediate influences.*

To estimate the PACF coefficient, we can use the following regression model:

$GDP_t = \alpha + \beta_1 GDP_{t-1} + \beta_2 GDP_{t-2} + \dots + \beta_k GDP_{t-k} + \epsilon_t$

where $\beta_k$ represents the **partial autocorrelation** at lag $k$, capturing only the **direct effect** of $GDP_{t-k}$ on $GDP_t$ after removing intermediate influences. The regression sequentially estimates PACF coefficients by including all prior lags and extracting the last coefficient in each model. This approach helps identify which lags have **direct predictive power**, which is crucial for **autoregressive model selection**.


We can also use the plot_pacf() function from the statsmodels library to visualize the PACF coefficients of a time series. 

```python
lags = range(1, 20) 
PACF_coeff = []

for shift_period in lags:
    GDP_t = GDP[shift_period:]

    GDP_t_lags = np.column_stack([GDP[shift_period-i:-i] for i in range(1, shift_period + 1)])
    GDP_t_lags_intercept = sm.add_constant(GDP_t_lags)

    model = sm.OLS(GDP_t, GDP_t_lags_intercept)
    result = model.fit()

    PACF_coeff.append(result.params[-1])  

plt.figure(figsize=(10, 5))
plt.stem(lags, PACF_coeff, basefmt="gray", markerfmt="o", linefmt="purple")
plt.axhline(0, color='red', linestyle='--', label="Zero Reference Line")

plt.xlabel("Lag Periods ($k$)")
plt.ylabel("OLS Estimated PACF Coefficients ($\\beta_k$)")
plt.title("OLS Estimated PACF for Lagged GDP Values")
plt.legend()
plt.show()
```

## Autoregressive (AR) Models

## Autoregressive (AR) Models

### AR(p) Model:
The general form of an **AR(p)** model expands on previous lags:
$$
X_t = \alpha + \phi_1 X_{t-1} + \phi_2 X_{t-2} + \dots + \phi_p X_{t-p} + \epsilon_t
$$
which can be concisely written as:
$$
X_t = \alpha + \sum_{i=1}^{p} \phi_i X_{t-i} + \epsilon_t
$$


### AR(1) Model:
$$
X_t = \alpha + \phi_1 X_{t-1} + \epsilon_t
$$

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
alpha = 0.5
phi = 0.7
n = 100
epsilon = np.random.normal(0, 1, n)

X = np.zeros(n)
X[0] = alpha + epsilon[0]

for t in range(1, n):
    X[t] = alpha + phi * X[t-1] + epsilon[t]

plt.plot(X, color='darkblue', label='AR(1) Process')  
plt.title('Simulated AR(1) Time Series') 
plt.xlabel('Time')  
plt.ylabel('Value')  
plt.legend() 
plt.grid(True)  
plt.show()  
```

#### Its ACF and PACF plots!

```python
from statsmodels.graphics.tsaplots import plot_acf
from statsmodels.graphics.tsaplots import plot_pacf

plot_acf(X, lags = 20, vlines_kwargs={"colors": "red"})
plt.title("ACF")
plt.show()

plot_pacf(X, lags=20, method="yw", vlines_kwargs={"colors": "red"})
plt.title("PACF")
plt.xticks()
plt.show()
```

### ADF test

```python
from statsmodels.tsa.stattools import adfuller

adf_result = adfuller(X)
print("adf_statistic", adf_result[0])
print("p_value", adf_result[1])
```

## Moving Average (MA) Models

### MA(q) Model:
Expanding the MA model to **q** lags:
$$
X_t = \alpha + \theta_1 \epsilon_{t-1} + \theta_2 \epsilon_{t-2} + \dots + \theta_q \epsilon_{t-q} + \epsilon_t
$$
which can be rewritten concisely as:
$$
X_t = \alpha + \sum_{j=1}^{q} \theta_j \epsilon_{t-j} + \epsilon_t
$$

### MA(1) Model:
$$
X_t = \alpha + \theta_1 \epsilon_{t-1} + \epsilon_t
$$

```python
n = 100
epsilon = np.random.normal(1, 5, n)

theta1 = 0.8
Y = np.zeros(n)

for t in range(1, n):
    Y[t] = epsilon[t] + theta1 * epsilon[t-1]

plt.figure(figsize=(10, 5))
plt.plot(Y, color='darkred', label='MA(1) Process')
plt.title('Simulated MA(1) Time Series')
plt.xlabel('Time')
plt.ylabel('Value')
plt.legend()
plt.grid(True)
plt.show()
```

#### Its ACF and PACF plots!

```python
plot_acf(Y, lags=20, vlines_kwargs={"colors": "red"})
plt.title("ACF")  

plot_pacf(Y, lags=20, method="yw", vlines_kwargs={"colors": "red"})
plt.title("PACF")  
plt.show()
```

### MA(2) Model:
$$
X_t = \alpha + \theta_1 \epsilon_{t-1} + \theta_2 \epsilon_{t-2} + \epsilon_t
$$

```python
np.random.seed(42)
n = 100
epsilon = np.random.normal(0, 1, n)

theta1 = 0.5
theta2 = 0.8
YY = np.zeros(n)

for t in range(2, n):
    YY[t] = epsilon[t] + theta1 * epsilon[t-1] + theta2 * epsilon[t-2]

plt.figure(figsize=(10, 5))
plt.plot(YY, color='blue', label='MA(2) Process')
plt.title('Simulated MA(2) Time Series')
plt.xlabel('Time')
plt.ylabel('Value')
plt.legend()
plt.grid(True)
plt.show()
```

#### Its ACF and PACF plots!

## How can we determine the order of an MA model?

#### The ACF of an MA model will show a cut-off after q lags. This means that the autocorrelations will be significant for the first q lags and then drop to zero.

#### The PACF for an MA model will decay gradually. Unlike AR models, the PACF is not as useful for determining the order of MA models, but it can still provide insights. Therefore, while the ACF is more definitive, examining the PACF can help confirm findings from the ACF.

## ARMA(p, q) Model

### ARMA Model:
$$
X_t = \alpha + \sum_{i=1}^{p} \phi_i X_{t-i} + \sum_{j=1}^{q} \theta_j \epsilon_{t-j} + \epsilon_t
$$

```python
alpha = 0.5  
phi = 0.7 
theta = 0.5  
n = 100  

epsilon = np.random.normal(0, 1, n)

XY = np.zeros(n)

for t in range(1, n):  # Start from t=1 due to AR(1)
    XY[t] = alpha + phi * XY[t-1]  + theta * epsilon[t-1]  + epsilon[t] 
    

plt.figure(figsize=(10, 5))
plt.plot(XY, color="green", linestyle="-", marker="o", markersize=4, alpha=0.7, label="ARMA(1,1) Simulation")
plt.title("Simulated ARMA(1,1) Process")
plt.xlabel("Time Step")
plt.ylabel("Value of XY")
plt.legend()
plt.show()
```

#### Its ACF and PACF plots!

```python
plot_acf(XY, lags=20, vlines_kwargs={"colors": "red"})
plt.title("ACF")  

plot_pacf(XY, lags=20, method="yw", vlines_kwargs={"colors": "red"})
plt.title("PACF")  
plt.show()
```

### Using statsmodels ArmaProcess

#### Its ACF and PACF plots!
